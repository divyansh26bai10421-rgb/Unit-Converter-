```python
km_to_miles=0.621371
miles_to_km=1.60934
kg_to_pounds=2.20462
pounds_to_kg=0.453592
```

```python
while True:
    print()
    print('''UNIT CONVERTER
1) Kilometres to Miles
2) Miles to Kilometres
3) Kilograms to Pounds
4) Pounds to Kilograms
5) Celsius to Fahrenheit
6) Fahrenheit to Celsius
7) Quit''')
    n=input('Enter your choice: ')
    if not n.isdigit():
        print('Please enter a number between 1 and 7')
        continue
    n=int(n)

    if n==7:
        print('Thank you, bye!')
        break
    elif n<1 or n>7:
        print('Please enter a number between 1 and 7')
        continue
    v=input('Enter the value: ')
    try:
        v=float(v)
    except ValueError:
        print('Please enter a number')
        continue
    if n==1:
        print(v, 'km =', round(v*km_to_miles, 2), 'miles')
    elif n==2:
        print(v, 'miles =', round(v*miles_to_km, 2), 'km')
    elif n==3:
        print(v, 'kg =', round(v*kg_to_pounds, 2), 'pounds')
    elif n==4:
        print(v, 'pounds =', round(v*pounds_to_kg, 2), 'kg')
    elif n==5:
        print(v, 'C =', round(v*9/5+32, 2), 'F')
    elif n==6:
        print(v, 'F =', round((v-32)*5/9, 2), 'C')
```
